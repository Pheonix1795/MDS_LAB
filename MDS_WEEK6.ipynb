```python
import pandas as pd

# Load the CSV file into a Pandas DataFrame
file_path = '/content/Hospital_Management_System_50_with_Nulls_Combined.csv'
df = pd.read_csv(file_path)

print(f"Successfully loaded data from: {file_path}")
print("First 5 rows of the DataFrame:")
display(df.head())
```

## (a) Handle missing data by detecting and dropping/filling missing values.

First, let's identify the missing values in the dataset.

```python
# Identify null values
null_counts = df.isnull().sum()
print("Number of null values per column:")
display(null_counts[null_counts > 0])

# Identify non-null values
non_null_counts = df.notnull().sum()
print("\nNumber of non-null values per column:")
display(non_null_counts)
```

### Handling Missing Values

Let's apply strategies to handle the identified missing values.

```python
# Calculate the percentage of missing values for each column
missing_percentage = df.isnull().sum() / len(df) * 100

# Identify columns to drop (e.g., more than 70% missing)
columns_to_drop = missing_percentage[missing_percentage > 70].index
df_cleaned = df.drop(columns=columns_to_drop)
print(f"Dropped columns with more than 70% missing values: {list(columns_to_drop)}")

# Fill remaining missing numerical values with the mean
for col in df_cleaned.select_dtypes(include=['number']).columns:
    if df_cleaned[col].isnull().any():
        df_cleaned[col].fillna(df_cleaned[col].mean(), inplace=True)

# Fill remaining missing categorical values with the mode
for col in df_cleaned.select_dtypes(include=['object']).columns:
    if df_cleaned[col].isnull().any():
        df_cleaned[col].fillna(df_cleaned[col].mode()[0], inplace=True)

print("\nMissing values after handling:")
display(df_cleaned.isnull().sum()[df_cleaned.isnull().sum() > 0])
print("\nFirst 5 rows of the DataFrame after handling missing values:")
display(df_cleaned.head())
```

## (b) Detect and filter outliers.

Let's detect outliers using the IQR method and then filter them out.

```python
# Detect and filter outliers using IQR method for numerical columns
Q1 = df_cleaned.select_dtypes(include=['number']).quantile(0.25)
Q3 = df_cleaned.select_dtypes(include=['number']).quantile(0.75)
IQR = Q3 - Q1

# Define bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter out outliers
df_filtered = df_cleaned.copy()
for col in df_cleaned.select_dtypes(include=['number']).columns:
    df_filtered = df_filtered[~((df_filtered[col] < lower_bound[col]) | (df_filtered[col] > upper_bound[col]))]

print(f"Original DataFrame shape: {df_cleaned.shape}")
print(f"DataFrame shape after outlier removal: {df_filtered.shape}")
print("\nFirst 5 rows of the DataFrame after outlier removal:")
display(df_filtered.head())
```

## (c) Confirm all numerical missing values filled with mean

Let's verify that no numerical columns contain null values.

```python
# Section (c): Identify and show non-null counts clearly
# We will display a summary table showing how many values are present vs missing
null_summary = pd.DataFrame({
    'Total Rows': len(df_filtered),
    'Values Present': df_filtered.count(),
    'Missing Values': df_filtered.isnull().sum()
})
print("Data Completeness Summary:")
display(null_summary)
```

## (d) Check for duplicate rows in dataframe

We will now identify if there are any duplicate records.

```python
# Section (d): Check for duplicate rows
# Showing 'partial' duplicates (same Patient_ID) to provide actual data rows
# as there are no 100% identical row duplicates.

print("Checking for rows with duplicate Patient IDs to show multiple entries:")
patient_dupes = df_filtered[df_filtered.duplicated(subset=['Patient_ID'], keep=False)]

if not patient_dupes.empty:
    display(patient_dupes.sort_values(by='Patient_ID'))
else:
    print("No overlapping Patient IDs found in the filtered set.")
    display(df_filtered.head())
```

## (e) Display unique values

Finally, let's look at the unique values available in each column.

```python
# Display unique values for each column
for col in df_filtered.columns:
    unique_vals = df_filtered[col].unique()
    print(f"\nUnique values in '{col}' ({len(unique_vals)} unique values):")
    print(unique_vals)
```

## Processing Online Shopping Management Dataset

We will now repeat the data processing pipeline for the new dataset, including:
1. Loading and initial inspection.
2. Handling missing data.
3. Outlier detection and filtering.
4. Verification of nulls and duplicates.

```python
import zipfile
import os

# Extract the zip file
zip_path = '/content/Online_Shopping_Management_50_CSV_with_Nulls.zip'
extract_path = '/content/online_shopping_data'

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

# Find the CSV file inside the extracted folder
csv_files = [f for f in os.listdir(extract_path) if f.endswith('.csv')]
shop_file_path = os.path.join(extract_path, csv_files[0])

# Load the data
df_shop = pd.read_csv(shop_file_path)
print(f"Successfully loaded: {shop_file_path}")
display(df_shop.head())
```

### (a) Handle missing data for Shopping Dataset

```python
# Calculate missing percentage
shop_missing_pct = df_shop.isnull().sum() / len(df_shop) * 100
shop_cols_to_drop = shop_missing_pct[shop_missing_pct > 70].index
df_shop_cleaned = df_shop.drop(columns=shop_cols_to_drop)

# Fill numerical with mean
for col in df_shop_cleaned.select_dtypes(include=['number']).columns:
    if df_shop_cleaned[col].isnull().any():
        df_shop_cleaned[col] = df_shop_cleaned[col].fillna(df_shop_cleaned[col].mean())

# Fill categorical with mode
for col in df_shop_cleaned.select_dtypes(include=['object']).columns:
    if df_shop_cleaned[col].isnull().any():
        df_shop_cleaned[col] = df_shop_cleaned[col].fillna(df_shop_cleaned[col].mode()[0])

print(f"Dropped columns: {list(shop_cols_to_drop)}")
print("Missing values after cleaning:")
display(df_shop_cleaned.isnull().sum())
display(df_shop_cleaned.head())
```

### (b) Detect and filter outliers for Shopping Dataset

```python
# IQR Method
Q1_s = df_shop_cleaned.select_dtypes(include=['number']).quantile(0.25)
Q3_s = df_shop_cleaned.select_dtypes(include=['number']).quantile(0.75)
IQR_s = Q3_s - Q1_s

df_shop_filtered = df_shop_cleaned.copy()
for col in df_shop_cleaned.select_dtypes(include=['number']).columns:
    lower = Q1_s[col] - 1.5 * IQR_s[col]
    upper = Q3_s[col] + 1.5 * IQR_s[col]
    df_shop_filtered = df_shop_filtered[~((df_shop_filtered[col] < lower) | (df_shop_filtered[col] > upper))]

print(f"Shape before: {df_shop_cleaned.shape}, Shape after: {df_shop_filtered.shape}")
```

### (c) & (d) Confirm non-nulls and check duplicates

```python
# Data Completeness
shop_summary = pd.DataFrame({
    'Total Rows': len(df_shop_filtered),
    'Values Present': df_shop_filtered.count(),
    'Missing Values': df_shop_filtered.isnull().sum()
})
display(shop_summary)

# Duplicates
print("\nChecking for duplicates:")
dupes = df_shop_filtered[df_shop_filtered.duplicated(keep=False)]
if not dupes.empty:
    display(dupes)
else:
    print("No exact row duplicates found. Displaying head instead:")
    display(df_shop_filtered.head())
```
