# 1. Create Pandas Series and Data Frame from various inputs

```python
import pandas as pd

# Creating a Series from a simple list
my_data_list = [5, 10, 15, 20]
my_series = pd.Series(my_data_list)
print("Simple Series:")
print(my_series)

# Creating a DataFrame from a dictionary
my_data_dict = {
    'Fruit': ['Apple', 'Banana', 'Cherry'],
    'Price': [1.2, 0.5, 2.5]
}
my_df = pd.DataFrame(my_data_dict)
print("\nSimple DataFrame:")
display(my_df)
```

# 2. Import CSV file to Pandas Data Frame

```python
# Loading the selected hospital data file
file_path = '/content/Hospital_Management_System_50_with_Nulls_Combined.csv'
hospital_df = pd.read_csv(file_path)
print("File loaded successfully!")
```

## (a) Visualize the first and last 10 records

```python
print("First 10 records:")
display(hospital_df.head(10))

print("\nLast 10 records:")
display(hospital_df.tail(10))
```

## (b) Get the shape, index and column details

```python
print("Rows and Columns count:", hospital_df.shape)
print("\nIndex of the data:", hospital_df.index)
print("\nColumn names:", hospital_df.columns.tolist())
```

## (c) Select/Delete records and columns based on conditions

```python
# Selection: Rows where 'Age' is greater than 30 (adjusting column name based on likely CSV content)
# Note: We check if 'Age' exists first
if 'Age' in hospital_df.columns:
    older_patients = hospital_df[hospital_df['Age'] > 30]
    print("Rows where Age > 30:")
    display(older_patients.head())

# Deletion: Drop the 'Phone' column if it exists
if 'Phone' in hospital_df.columns:
    df_no_phone = hospital_df.drop(columns=['Phone'])
    print("\nData after dropping 'Phone' column:")
    display(df_no_phone.head())
```

## (d) Perform ranking and sorting operations

```python
# Sorting by the first column in the DataFrame
first_col = hospital_df.columns[0]
df_sorted = hospital_df.sort_values(by=first_col, ascending=True)
print(f"Sorted by {first_col}:")
display(df_sorted.head())

# Ranking values in the second column
second_col = hospital_df.columns[1]
hospital_df['Rank_Value'] = hospital_df[second_col].rank()
print(f"\nData with Rank column based on {second_col}:")
display(hospital_df.head())
```

## (e) Statistical operations on the given columns

```python
print("Statistical summary for numeric data:")
display(hospital_df.describe())
```

## (f) Count and uniqueness of categorical values

```python
# Finding a categorical column (text-based)
cat_col = hospital_df.select_dtypes(include=['object']).columns[0]
print(f"Column analyzed: {cat_col}")
print(f"Number of unique items: {hospital_df[cat_col].nunique()}")
print("\nCounts for each item:")
print(hospital_df[cat_col].value_counts())
```

## (g) Rename single/multiple columns

```python
# Renaming the first column to 'ID_Number' and second to 'Main_Entry'
renamed_df = hospital_df.rename(columns={
    hospital_df.columns[0]: 'ID_Number',
    hospital_df.columns[1]: 'Main_Entry'
})
print("Columns after renaming:")
print(renamed_df.columns.tolist()[:5])
```
