```python
import numpy as np
a = np.array([1,2,3,4,5,6])
b = np.array([6,7,8,9,10,11])
print("Elements of a : ", a)
print("Elements of b : ", b)
print("Indexing : ", a[2])
print("Slicing : ", a[1:3])
print("Reshaping : ", np.reshape(a,(3,2)))
print("Joining : ", np.concatenate((a,b)))
print("Splitting : ", np.split(a,2))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
Data = {
    'Class': ('A','B','C','D','E'),
    'Score': (99,75,81,40,20)
}
df = pd.DataFrame(Data)

plt.bar(df['Class'],df['Score'])
plt.show()

plt.barh(df['Class'],df['Score'])
plt.show()
```
